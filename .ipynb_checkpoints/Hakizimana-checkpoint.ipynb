```python
# Then import the required libraries:

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Now load the dataset:

df = pd.read_csv("Student_Dataset_Dirty.csv")

# Task1: Display the first five rows, dataset shape, column types and descriptive statistics.

print("FIRST FIVE ROWS")
print(df.head())

print("\nDATASET SHAPE")
print(df.shape)

print("\nCOLUMN TYPES")
print(df.dtypes)

print("\nDESCRIPTIVE STATISTICS")
print(df.describe(include="all"))
```

```python
# Task2. Count missing values in every column.
print("\nMISSING VALUES BEFORE CLEANING")
print(df.isnull().sum())
```

```python
# Task3. Identify duplicate student records and remove them using Student_ID as the key.
duplicates = df[df.duplicated("Student_ID", keep=False)]

print("\nDUPLICATE STUDENT RECORDS")
print(duplicates)

print("\nDUPLICATE STUDENT IDs")
print(duplicates["Student_ID"].drop_duplicates().tolist())

df = df.drop_duplicates(
    subset="Student_ID",
    keep="first"
)
```

```python

#Task4.Remove leading and trailing spaces from text fields.

text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()

```

```python
#Task5. Standardize Gender, Campus, Internet_Access and Preferred_Learning_Mode.
# Standardize Gender
df["Gender"] = (
    df["Gender"]
    .str.lower()
    .replace({
        "m": "Male",
        "male": "Male",
        "f": "Female",
        "female": "Female"
    })
)

# Standardize Campus
df["Campus"] = (
    df["Campus"]
    .str.lower()
    .str.title()
)

# Standardize Internet Access
df["Internet_Access"] = (
    df["Internet_Access"]
    .str.lower()
    .replace({
        "y": "Yes",
        "yes": "Yes",
        "n": "No",
        "no": "No"
    })
)

# Standardize Preferred Learning Mode
df["Preferred_Learning_Mode"] = (
    df["Preferred_Learning_Mode"]
    .str.lower()
    .replace({
        "online": "Online",
        "blended": "Blended",
        "face to face": "Face-to-Face",
        "face-to-face": "Face-to-Face"
    })
)
```

```python
#Task6: Find values outside sensible ranges for age, study hours, attendance and marks.
valid_ranges = {
    "Age": (15, 100),
    "Study_Hours_Per_Week": (0, 168),
    "Attendance_Percent": (0, 100),
    "Previous_Score": (0, 100),
    "Final_Mark": (0, 100)
}

print("\nINVALID VALUES")

for column, (minimum, maximum) in valid_ranges.items():

    invalid = (
        (df[column] < minimum) |
        (df[column] > maximum)
    )

    print(
        column,
        ":",
        df.loc[invalid, column].tolist()
    )
    # Convert invalid values to NaN
    df.loc[invalid, column] = np.nan
```

```python
#Task7: Treat invalid values as missing and replace missing numerical values with the median of valid observations.

# Replace missing numerical values with median
for column in valid_ranges:

    median_value = df[column].median()

    df[column] = df[column].fillna(
        median_value
    )
```

```python
# Task8. Convert Submission_Date to one consistent YYYY-MM-DD format.

df["Submission_Date"] = pd.to_datetime(
    df["Submission_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

df["Submission_Date"] = (
    df["Submission_Date"]
    .dt.strftime("%Y-%m-%d")
)
```

```python
#Task9. Remove any column that does not contribute to the analysis.

df = df.drop(
    columns=["Temporary_Note"],
    errors="ignore"
)
```

```python
#Task10. Confirm the final number of rows and verify that no missing values remain.
print("\nFINAL DATASET SHAPE")
print(df.shape)

print("\nFINAL MISSING VALUES")
print(df.isnull().sum())

if df.isnull().sum().sum() == 0:
    print("\nSUCCESS: No missing values remain.")
else:
    print("\nWARNING: Missing values still exist.")
```

```python
#Task11. Save the cleaned data as Student_Dataset_Cleaned.csv.
df.to_csv(
    "Student_Dataset_Cleaned.csv",
    index=False
)

print(
    "\nCleaned dataset saved as "
    "Student_Dataset_Cleaned.csv"
)
```

```python
# REQUIRED VISUALIZATION 1
# Histogram of Final_Mark
# ==========================================================



plt.figure(figsize=(8, 5))

plt.hist(
    df["Final_Mark"],
    bins=8,
    edgecolor="gold"
)

plt.title("Histogram of Final Mark")
plt.xlabel("Final Mark")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
# REQUIRED VISUALIZATION 2
# Bar chart of average Final_Mark by Gender
# ==========================================================

gender_average = (
    df.groupby("Gender")["Final_Mark"]
    .mean()
)

print("\nAVERAGE FINAL MARK BY GENDER")
print(gender_average)

plt.figure(figsize=(7, 5))

plt.bar(
    gender_average.index,
    gender_average.values,
    edgecolor="yellow"
)

plt.title("Average Final Mark by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Mark")

plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

```python
# REQUIRED VISUALIZATION 3
# Scatter plot of Study_Hours_Per_Week against Final_Mark, distinguished by Campus
# ================================================================================

plt.figure(figsize=(8, 5))

for campus in df["Campus"].unique():

    campus_data = df[
        df["Campus"] == campus
    ]

    plt.scatter(
        campus_data["Study_Hours_Per_Week"],
        campus_data["Final_Mark"],
        label=campus,
        s=70
    )

plt.title(
    "Study Hours per Week vs Final Mark by Campus"
)

plt.xlabel("Study Hours per Week")
plt.ylabel("Final Mark")

plt.legend(title="Campus")

plt.tight_layout()
plt.show()
```

```python
# Two written findings supported by the analysis
#===============================================

correlation = (
    df["Study_Hours_Per_Week"]
    .corr(df["Final_Mark"])
)

print("\nTWO WRITTEN FINDINGS")

print(
    f"1. Study hours and final marks show a positive "
    f"relationship, with a correlation of approximately "
    f"{correlation:.2f}."
)

print(
    f"2. The average Final Mark was "
    f"{gender_average['Female']:.2f} for Female students "
    f"and {gender_average['Male']:.2f} for Male students."
)

```
