```python
import pandas as pd
df =pd.read_csv("/content/house_price_prediction_dataset.csv")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
print(f"Column names: {list(df.columns)}")
df.head(10)
```

```python
print("--- 2.1 Basic statistics ---")
display(df.describe())
```

```python
print("---2.2 Data types---")
print(df.dtypes)
```

```python
print("---2.3 Missing values per solumn----")
print(df.isna().sum())
```

```python
print("--- 2.4 Duplicate rows ---")
print(f"Duplicate rows: {df.duplicated().sum()}")

print("--- 2.5 Unique values ---")
for col in df.columns:  # Loop over every column
    print(f"{col}: {df[col].nunique()} unique values")
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Check how many rows remain
print(f"Rows after removing duplicates: {len(df)}")
```

```python
print("--- 6.1 Detecting outliers using IQR ---")

numeric_cols = df.select_dtypes(include=['number']).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    print(f"{col}: {outliers} outliers")
```

```python
print("--- 7.1 Removing outliers ---")

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

print(f"Rows after removing outliers: {len(df)}")
```

```python
print("--- 8.1 Checking categorical values ---")

categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    print(f"\n{col}")
    print(df[col].value_counts())
```

```python
print("--- 9.1 Final data quality check ---")

print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
print("--- 11. Saving cleaned dataset ---")

df.to_csv("/content/cleaned_house_price_prediction_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
